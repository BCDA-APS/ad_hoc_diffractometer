```python
import hklpy2
import numpy as np

np.set_printoptions(precision=4, suppress=True)

fourc = hklpy2.creator(geometry="E4CV", name="fourc", reals="th chi phi tth".split())
fourc.wh()
```

```python
# in SPEC, these are the default sample, reflections, and wavelength
"""text
  Primary Reflection (at lambda 1.54):
          tth th chi phi = 60 30 0 0 
                   H K L = 1 0 0

  Secondary Reflection (at lambda 1.54):
          tth th chi phi = 60 30 0 -90 
                   H K L = 0 1 0

  Lattice Constants (lengths / angles):
              real space = 1.54 1.54 1.54 / 90 90 90
        reciprocal space = 4.08 4.08 4.08 / 90 90 90
"""
fourc.beam.wavelength.put(1.54)
fourc.add_sample("SPEC default sample", 1.54)
r1 = fourc.add_reflection((1, 0, 0), dict(tth=60, th=30, chi=0, phi=0))
r2 = fourc.add_reflection((0, 1, 0), dict(tth=60, th=30, chi=0, phi=-90))
fourc.core.calc_UB(r1, r2)
```

```python
"""text
170.FOURCC> pa

Four-Circle Geometry, Omega equals zero (mode 0)
Sector 0
...
"""
fourc.core.mode = "constant_omega"  # be sure to set th to zero
```

```python
# C Sun Dec 13 20:42:09 2020.  sample is sapphire with 001 parallel to phi axis.
# C Sun Dec 13 20:42:35 2020.  g_aa reset from 1.54 to 4.785.
# C Sun Dec 13 20:42:35 2020.  g_bb reset from 1.54 to 4.785.
# C Sun Dec 13 20:42:35 2020.  g_cc reset from 1.54 to 12.991.
# C Sun Dec 13 20:42:35 2020.  g_ga reset from 90 to 120.

fourc.add_sample("sapphire (001) parallel to phi", a=4.785, c=12.991, gamma=120)
```

```python
# C Sun Dec 13 20:44:12 2020.  g_lambda0 reset from 1.54 to 1.5498.
fourc.beam.wavelength.put(1.549802558)
```

```python
# C Sun Dec 13 20:43:36 2020.  Now set up a fake orientation matrix that will approximate the actual orientation.
# C Sun Dec 13 20:44:12 2020.  g_h0 reset from 1 to 0.
# C Sun Dec 13 20:44:12 2020.  g_l0 reset from 0 to 6.
# C Sun Dec 13 20:44:12 2020.  g_u00 reset from 60 to 41.9419.
# C Sun Dec 13 20:44:12 2020.  g_u01 reset from 30 to 20.97.
# C Sun Dec 13 20:44:12 2020.  g_u02 reset from 0 to 90.
# C Sun Dec 13 20:44:12 2020.  g_lambda0 reset from 1.54 to 1.5498.
# C Sun Dec 13 20:44:38 2020.  that was the 006 at roughly where it should be.

# G1 4.785 4.785 12.991 90 90 120 1.516237713 1.516237713 0.483656786 90 90 60 0 0 6 1 0 0 41.94188 20.97 90 0 0 0 60 30 0 0 0 0 1.549802558 1.549802558 0 0
r1 = fourc.add_reflection(
    (0, 0, 6), dict(tth=41.94188, th=20.97, chi=90, phi=0), wavelength=1.549802558
)
```

```python
# C Sun Dec 13 20:46:09 2020.  I do not want to go to phi=-90.
# C Sun Dec 13 20:49:49 2020.  detector is an APD about 1 m from sample, behind a fairly open slit.
# C Sun Dec 13 20:49:59 2020.  so there is a lot of background.
# C Sun Dec 13 20:50:08 2020.  try to find 006 reflection.
# ...  after scanning all real axes to obtain precise peak positions, iterative
# C Sun Dec 13 21:07:31 2020.  g_u00 reset from 41.9419 to 41.9394.
# C Sun Dec 13 21:07:31 2020.  g_u01 reset from 20.97 to 20.3654.
# C Sun Dec 13 21:07:31 2020.  g_u02 reset from 90 to 89.32.
# C Sun Dec 13 21:08:40 2020.  I just entered the first reflection in the orientation matrix so spec at least knows which way is up.
# C Sun Dec 13 21:09:45 2020.  I think I saw something at th ~ 15 so will see if that is real.
# ...
# C Sun Dec 13 22:01:47 2020.  I optimized the signal from this peak with scans of th, chi, and tth.
# C Sun Dec 13 22:02:19 2020.  I did not scan phi because that would be redundant.
# C Sun Dec 13 22:02:31 2020.  now enter this reflection as the second in the orientation matrix.
# C Sun Dec 13 22:02:40 2020.  g_l1 reset from 0 to 4.
# C Sun Dec 13 22:02:40 2020.  g_u10 reset from 60 to 35.3924.
# C Sun Dec 13 22:02:40 2020.  g_u11 reset from 30 to 17.6428.
# C Sun Dec 13 22:02:40 2020.  g_u12 reset from 0 to 50.8925.
# C Sun Dec 13 22:02:40 2020.  g_u13 reset from 0 to 29.95.
# Update r1, define r2 (after locating & refininig peak positions)
r1 = fourc.add_reflection(
    (0, 0, 6),
    dict(tth=41.93937, th=20.36538, chi=89.32, phi=0),
    wavelength=1.549802558,
    replace=True,
)
r2 = fourc.add_reflection(
    (1, 0, 4),
    dict(tth=35.39238, th=17.64281, chi=50.89250, phi=29.95),
    wavelength=1.549802558,
    replace=True,
)
```

```python
# C Sun Dec 13 22:03:22 2020.  swap orientations to make this the primary one.
# C Sun Dec 13 22:03:23 2020.  Orientation vectors swapped.
fourc.core.calc_UB(r2, r1)
```

```python
"""text
120.FOURCC> wh

H K L =  1.0027  4.0939e-07  4.0109
Alpha = 14.636  Beta = 13.05  Azimuth = -91.31
Omega = 0  Lambda = 1.5498

  TwoTheta      Theta        Chi        Phi 
  35.39238   17.64281   50.89250   29.95000 
"""
# C Sun Dec 13 22:04:05 2020.  the apparent HKL position is a little off; maybe I do not have the optimal lattice parameters.
# C Sun Dec 13 22:08:52 2020.  now should be able to br to any Bragg reflection and be close.
fourc.tth.move(35.39238)
fourc.th.move(17.64281)
fourc.chi.move(50.89250)
fourc.phi.move(29.95)
fourc.wh()
```

```python
# to keep omega=0 mode, move the 'th' axis back to zero
fourc.th.move(0)
print(f"{fourc.core.mode=}")
fourc.wh()
```
